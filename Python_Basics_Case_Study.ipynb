```python
# Bank Account Manager

account_holder = "Gourinanda"
account_number = 123456789
balance = 5000.0

print("Bank Account Created Successfully!")
print("Account Holder:", account_holder)
print("Account Number:", account_number)
print("Initial Balance: Rs.", balance)
```

```python
def deposit(amount):
    global balance

    if amount > 0:
        balance = balance + amount
        print("Deposit Successful!")
        print("Deposited Amount: Rs.", amount)
        print("Updated Balance: Rs.", balance)
    else:
        print("Please enter a valid amount.")


# Testing deposit function
deposit(2000)
```

```python
def withdraw(amount):
    global balance

    if amount > 0:
        if amount <= balance:
            balance = balance - amount
            print("Withdrawal Successful!")
            print("Withdrawn Amount: Rs.", amount)
            print("Remaining Balance: Rs.", balance)
        else:
            print("Insufficient Balance!")
    else:
        print("Please enter a valid amount.")


# Testing withdrawal function
withdraw(1500)
```

```python
def check_balance():
    print("Account Holder:", account_holder)
    print("Account Number:", account_number)
    print("Available Balance: Rs.", balance)


# Testing balance enquiry
check_balance()
```

```python
withdraw(10000)
```

```python
# Weather Data Analyzer

region = "Kerala"

temperature = [28, 30, 32, 29, 27]
humidity = [75, 70, 65, 80, 85]

print("Weather Data for:", region)
print("Temperature:", temperature)
print("Humidity:", humidity)
```

```python
def check_weather(temp):

    if temp >= 30:
        print("Temperature:", temp, "°C - Sunny Weather")

    elif temp >= 25:
        print("Temperature:", temp, "°C - Cloudy Weather")

    else:
        print("Temperature:", temp, "°C - Cool Weather")


# Checking weather for each day

for temp in temperature:
    check_weather(temp)
```

```python
# Function to calculate average temperature

def calculate_average_temperature():
    average_temp = sum(temperature) / len(temperature)
    print("Average Temperature:", average_temp, "°C")


# Function to calculate average humidity

def calculate_average_humidity():
    average_hum = sum(humidity) / len(humidity)
    print("Average Humidity:", average_hum, "%")


# Calling the functions

calculate_average_temperature()
calculate_average_humidity()
```
